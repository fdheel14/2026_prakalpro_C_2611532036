## 2.1 Dasar Teori

### 1. Struktur Percabangan

Struktur percabangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk membuat program mengambil keputusan berdasarkan kondisi tertentu. Dengan percabangan, program dapat menjalankan perintah yang berbeda sesuai dengan keadaan atau nilai yang diberikan oleh pengguna. Dalam Python, struktur percabangan yang umum digunakan antara lain if, if-else, if-elif-else, multi-if, dan match-case.

### 2. Struktur if

if digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar. Jika kondisi bernilai salah, perintah di dalam if tidak akan dijalankan.

Contoh penggunaan if adalah:

nilai_2036 = 88

if nilai_2036 >= 80:

    print("Lulus")

### 3. Struktur if-else

if-else digunakan ketika terdapat dua kemungkinan hasil. Jika kondisi pada if benar, program menjalankan perintah pada bagian if. Jika salah, program menjalankan perintah pada bagian else.

Contoh penggunaan if-else adalah:

nilai_2036 = 55

if nilai_2036 >= 80: 

    print("Lulus") 

else: 

    print("Tidak Lulus")

### 4. Struktur if-elif-else

if-elif-else digunakan untuk memeriksa beberapa kondisi secara berurutan. Program akan menjalankan blok perintah pada kondisi yang pertama kali bernilai benar.

Contoh penggunaan if-elif-else adalah:

nilai_2036 = 80

if nilai_2036 >= 90: 

    print("Sangat Baik") 

elif nilai_2036 >= 75:

   print("Baik") 
else: 

 print("Kurang")

### 5. Multi-if

Multi-if adalah penggunaan beberapa pernyataan if secara terpisah dalam satu program. Berbeda dengan if-elif-else, setiap kondisi pada multi-if diperiksa secara independen. Artinya, lebih dari satu kondisi dapat bernilai benar dan semua perintah dari kondisi yang terpenuhi dapat dijalankan.

### 6. Operator Perbandingan dan Logika

Operator perbandingan dan operator logika digunakan dalam struktur percabangan untuk menentukan dan menggabungkan kondisi tertentu. Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan nilai True atau False. Operator yang umum digunakan yaitu == (sama dengan), != (tidak sama dengan), > (lebih besar), < (lebih kecil), >= (lebih besar atau sama dengan), dan <= (lebih kecil atau sama dengan). Sementara itu, operator logika digunakan untuk menggabungkan atau memodifikasi beberapa kondisi. Operator logika dalam Python terdiri dari and, or, dan not. Penggunaan kedua jenis operator tersebut memungkinkan program membuat keputusan berdasarkan kondisi yang lebih kompleks.

### 7. Input dan Konversi Data

Fungsi input() digunakan untuk menerima data yang dimasukkan oleh pengguna melalui keyboard. Secara default, hasil dari input() berupa data bertipe string.

Untuk mendapatkan tipe data tertentu, input dapat dikonversi menggunakan fungsi seperti int() dan float(). Fungsi int() digunakan untuk mengubah data menjadi bilangan bulat, sedangkan float() digunakan untuk mengubah data menjadi bilangan desimal.

Contohnya:

umur_2036 = int(input("Input umur anda: "))

### 8. Penggunaan strip(), lower(), dan [0]

Metode. strip() digunakan untuk menghapus spasi yang terdapat di awal atau akhir teks. Metode. lower() digunakan untuk mengubah seluruh huruf dalam teks menjadi huruf kecil. Kedua metode tersebut dapat digunakan untuk membuat input pengguna lebih mudah dibandingkan dengan kondisi yang telah ditentukan.

Contohnya:

input_member_2036 = input("Apakah Anda member? (y/t): ").strip().lower()

[0] dapat digunakan untuk mengambil karakter pertama dari sebuah string. Pada program pemeriksaan SIM C, [0] digunakan untuk mengambil karakter pertama dari jawaban pengguna, misalnya y atau t.

### 9. match-case

match-case merupakan struktur percabangan pada Python yang digunakan untuk mencocokkan suatu nilai dengan beberapa pola atau pilihan yang telah ditentukan. Struktur ini dapat digunakan sebagai alternatif dari if-elif-else ketika program memiliki banyak pilihan berdasarkan nilai tertentu. match digunakan untuk menentukan nilai yang akan diperiksa, sedangkan case digunakan untuk menentukan pola atau nilai yang akan dicocokkan.

Contoh Penggunaan:

bulan_2036 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_2036:

    case 1:

        print("Januari")

    case 2:

            print("Februari")

    case 3:

            print("Maret")

### 10. Operasi Perhitungan Diskon

Dalam program perhitungan diskon, operator aritmatika digunakan untuk menghitung nominal diskon dan total pembayaran. Persentase diskon diubah terlebih dahulu menjadi bentuk desimal dengan membaginya dengan 100.

Rumus yang digunakan adalah:

Nominal Diskon = Total Belanja × (Total Diskon / 100)

Total Bayar = Total Belanja - Nominal Diskon

Pada program multi-if, beberapa persentase diskon dapat dijumlahkan terlebih dahulu apabila pengguna memenuhi beberapa syarat diskon.

### 11. Konversi Nilai Angka Menjadi Nilai Huruf

Struktur if-elif-else dapat digunakan untuk mengelompokkan nilai angka ke dalam kategori tertentu. Pada program yang dibuat, nilai angka dikonversi menjadi nilai huruf berdasarkan rentang nilai.

Contoh:

nilai_2036 = int(input("Inputkan nilai angka= "))

if nilai_2036 >= 81:

    print("A")

elif nilai_2036 >= 70:

    print("B")

elif nilai_2036 >= 60:

    print("C")

elif nilai_2036 >= 50:

    print("D")

else:

    print("E")

### 2.2.4 Program "multi_if2_2611532036.py"

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_1234
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2036 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2036 = input("Apakah anda member? (y/t): ").strip().lower()
is_member_2036 = input_member_2036 in ["y","ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2036 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2036 = input_promo_2036 in ["y","ya"]

total_diskon_persen_2036 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2036 > 1000000:
    total_diskon_persen_2036 += 10 # Diskon belanja besar

if is_member_2036:
    total_diskon_persen_2036 += 5 # Diskon member

if kode_promo_valid_2036:
    total_diskon_persen_2036 += 15 # Diskon voucher

# Menghitung nominal diskon dan total bayar
nominal_diskon_2036 = total_belanja_2036 * (total_diskon_persen_2036 / 100)
total_bayar_2036 = total_belanja_2036 - nominal_diskon_2036

# Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_2036}% (Rp {nominal_diskon_2036:,.0f})")
print(f"Total Bayar : Rp {total_bayar_2036:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_2036}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promosi valid
```

### 2.2.7 Program "match_case_2611532036.py"

```python
# Buat file dengan nama match_case1.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_2036 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_2036:
    case 1:
        print("Januari")
    case 2:
            print("Februari")
    case 3:
            print("Maret")
    case 4:
            print("April")
    case 5:
            print("Mei")
    case 6:
            print("Juni")
    case 7:
            print("Juli")
    case 8:
            print("Agustus")
    case 9:
            print("September")
    case 10:
            print("Oktober")
    case 11:
            print("November")
    case 12:
            print("Desember")
    case _:
        print("Angka tidak valid")
```

## Tugas Praktikum

Soal :

Sebuah wahana rekreasi petualangan membutuhkan sistem loket berbasis konsol untuk menentukan jenis wahana, memverifikasi kelaikan pengunjung dengan kondisi majemuk (logical operators), menghitung diskon akumulatif menggunakan multi-if terpisah, serta mencetak audit status operasional.

Kode Jawaban :

a. Kode Program "Tugas4_2611532036.py"

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. Input Data Pengunjung

nama_pengunjung_2036 = input("Masukkan Nama Pengunjung            : ")
umur_2036 = int(input("Input Umur Anda                     : "))
sim_2036 = input("Apakah Anda Sudah Punya SIM C (y/t) : ").strip().lower()

# 2. Pemilihan Wahana Menggunakan match-case

print("\nPilihan Paket Wahana (1-5):")
print(" 1. Safari Rimba         (Rp 50,000)")
print(" 2. Arung Jeram          (Rp 75,000)")
print(" 3. Motor ATV Ekstrim    (Rp 120,000)")
print(" 4. Roller Coaster Kilat (Rp 100,000)")
print(" 5. All-Access VIP       (Rp 220,000)")

paket_2036 = int(input("Masukkan nomor paket(1-5)      : "))
match paket_2036:
    case 1:
        nama_wahana_2036 = "Wahana Safari Rimba"
        harga_satuan_2036 = 50000
    case 2:
        nama_wahana_2036 = "Wahana Arung Jeram"
        harga_satuan_2036 = 75000
    case 3:
        nama_wahana_2036 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2036 = 120000
    case 4:
        nama_wahana_2036 = "Wahana Roller Coaster Kilat"
        harga_satuan_2036 = 100000
    case 5:
        nama_wahana_2036 = "Wahana All-Access VIP"
        harga_satuan_2036 = 220000
    case _:
        print("Paket tidak valid")
        raise SystemExit

jumlah_tiket_2036 = int(input("Masukkan jumlah tiket          : "))
if jumlah_tiket_2036 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

member_2036 = input("Apakah anda member? (y/t)      : ").strip().lower()
is_member_2036 = member_2036 in ["y","ya","yes"]

promo_2036 = input("Apakah kode promo valid? (y/t) : ").strip().lower()
kode_promo_valid_2036 = promo_2036 in ["y","ya","yes"]

# 3. Validasi Izin Kendali Wahana Menggunakan if - elif - else dan Operator Logika (and, !=)

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_2036 == 3 and umur_2036 >= 17 and sim_2036 == 'y':
    status_akses_2036 = "Anda sudah dewasa dan boleh mengendarai ATV sendiri."
elif paket_2036 == 3 and umur_2036 >= 17 and sim_2036 != 'y':
    status_akses_2036 = "Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur)."
elif paket_2036 == 3 and umur_2036 < 17 and sim_2036 == 'y':
    status_akses_2036 = "Identitas tidak valid: Belum cukup umur memiliki SIM"
else:
    status_akses_2036 = "Anda belum cukup umur dan tidak boleh bawa motor ATV"

print(f"Status Akses: {status_akses_2036}")

# 4. Akumulasi Diskon Bertingkat Menggunakan Multi-IF Terpisah

subtotal_2036 = harga_satuan_2036 * jumlah_tiket_2036

total_diskon_persen_2036 = 0

# Diskon belanja besar
if subtotal_2036 >= 200000:
   total_diskon_persen_2036 += 10

# Diskon member
if is_member_2036:
   total_diskon_persen_2036 += 5

# Diskon voucher promo
if kode_promo_valid_2036:
   total_diskon_persen_2036 += 15

# Diskon tambahan rombongan
if jumlah_tiket_2036 >= 5:
   total_diskon_persen_2036 += 5

# 5. Evaluasi Kelulusan Audit Menggunakan if - else

# Perhitungan nominal diskon dan total bayar
nominal_diskon_2036 = subtotal_2036 * (total_diskon_persen_2036 / 100)

total_bayar_2036 = subtotal_2036 - nominal_diskon_2036

# Evaluasi total bayar
if total_bayar_2036 > 300000:
    catatan_layanan_2036 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
else:
    catatan_layanan_2036 = "Terima kasih telah berkunjung."

# 6. Rincian Pembayaran dengan f-string format desimal (:,.0f).

print("\n--- RINCIAN PEMBAYARAN ---")
print(f"Nama Pengunjung   : {nama_pengunjung_2036}")
print(f"Wahana            : {nama_wahana_2036}")
print(f"Harga Satuan      : Rp{harga_satuan_2036:,.0f}")
print(f"Jumlah Tiket      : {jumlah_tiket_2036}")
print(f"Subtotal Belanja  : Rp{subtotal_2036:,.0f}")
print(f"Total Diskon      : {total_diskon_persen_2036}% " f"(Rp{nominal_diskon_2036:,.0f})")
print(f"Total Bayar       : Rp{total_bayar_2036:,.0f}")
print(f"Catatan Layanan   : {catatan_layanan_2036}")

print("\nProgram Selesai")
```

Penjelasan : 

Baris 1–5 berfungsi untuk menampilkan judul program serta meminta input data pengunjung berupa nama, umur, kepemilikan SIM C, dan jumlah tiket.

Baris 8–9 berfungsi untuk melakukan validasi jumlah tiket menggunakan if tunggal. Jika jumlah tiket kurang dari atau sama dengan 0, program akan menampilkan peringatan bahwa kuota tiket tidak valid.

Baris 13–18 berfungsi untuk menampilkan daftar pilihan paket wahana beserta harga masing-masing paket.

Baris 20 berfungsi untuk menerima input nomor paket wahana yang dipilih oleh pengguna dan mengubah input tersebut menjadi tipe data integer.

Baris 22–45 berfungsi untuk menentukan nama wahana dan harga satuan berdasarkan nomor paket yang dipilih menggunakan struktur match-case. case 1 sampai case 5 digunakan untuk menentukan lima pilihan wahana yang tersedia, sedangkan case _ digunakan untuk menangani input paket yang tidak valid.

Baris 47–48 berfungsi untuk menghentikan program apabila pengguna memasukkan nomor paket yang tidak tersedia.

Baris 53–82 berfungsi untuk melakukan pemeriksaan kelayakan pengunjung dalam menggunakan wahana. Pada paket Motor ATV Ekstrim, kondisi diperiksa berdasarkan umur dan kepemilikan SIM C menggunakan operator logika and dan !=. Untuk paket selain ATV, kelayakan diperiksa berdasarkan umur minimal 10 tahun.

Baris 84 berfungsi untuk menampilkan hasil pemeriksaan kelayakan pengunjung berdasarkan kondisi yang telah dievaluasi sebelumnya.

Baris 88–92 berfungsi untuk meminta input status keanggotaan pengguna dan validitas kode promo. Input kemudian diubah menjadi huruf kecil menggunakan .lower() agar dapat dibandingkan dengan kondisi yang telah ditentukan.

Baris 97 berfungsi untuk menghitung subtotal belanja dengan mengalikan harga satuan wahana dengan jumlah tiket yang dibeli.

Baris 102–117 berfungsi untuk menghitung total diskon menggunakan beberapa if yang berdiri secara terpisah. Diskon diberikan berdasarkan subtotal belanja, status member, kode promo, dan jumlah tiket. Karena menggunakan multi-if, beberapa diskon dapat diperoleh dan dijumlahkan secara bersamaan.

Baris 122–127 berfungsi untuk menghitung nominal diskon dan total pembayaran. Nominal diskon diperoleh dari subtotal dikalikan persentase diskon, kemudian total pembayaran diperoleh dengan mengurangi subtotal dengan nominal diskon.

Baris 132–139 berfungsi untuk melakukan evaluasi akhir menggunakan if-else. Jika total pembayaran lebih dari Rp300.000, pengguna mendapatkan informasi bahwa dirinya berhak memperoleh souvenir gratis. Jika tidak, program menampilkan ucapan terima kasih.

Baris 144–159 berfungsi untuk menampilkan seluruh rincian transaksi, yaitu nama pengunjung, nama wahana, harga satuan, jumlah tiket, subtotal, total diskon, total pembayaran, dan catatan layanan. Format :,.0f digunakan untuk menampilkan nominal angka dengan pemisah ribuan tanpa angka desimal.

Baris 161 berfungsi untuk menampilkan pesan bahwa program telah selesai dijalankan.
