## 2.1 Dasar Teori

### 1. Python

Python merupakan bahasa pemrograman tingkat tinggi yang dirancang dengan sintaks yang sederhana dan mudah dipahami. Python banyak digunakan dalam berbagai bidang, seperti pengembangan perangkat lunak, pengolahan data, kecerdasan buatan, dan pembuatan program sederhana.

Pada praktikum ini, Python digunakan untuk membuat program yang dapat menghasilkan pola Jam Pasir Kristal Palindromik Berbingkai secara dinamis berdasarkan nilai yang dimasukkan oleh pengguna.

### 2. Perulangan for

Perulangan for merupakan struktur kontrol pada Python yang digunakan untuk menjalankan suatu blok kode secara berulang berdasarkan urutan atau rentang nilai tertentu. Perulangan ini dapat digunakan ketika jumlah proses pengulangan telah ditentukan atau dapat diperoleh dari suatu nilai.

Bentuk umum perulangan for adalah:
for variabel in range(nilai): 
    perintah

Pada program praktikum ini, perulangan for digunakan untuk menentukan jumlah baris pola, mencetak angka, menghasilkan spasi, serta membentuk karakter pada bagian bingkai.

### 3. Fungsi range()

Fungsi range() digunakan untuk menghasilkan urutan bilangan yang dapat dimanfaatkan sebagai pengendali perulangan. Fungsi ini dapat memiliki parameter batas akhir, nilai awal dan batas akhir, serta nilai awal, batas akhir, dan langkah perubahan.

### 4. Nested for

Nested for atau perulangan bersarang merupakan perulangan yang terdapat di dalam perulangan lainnya. Perulangan luar biasanya digunakan untuk mengatur proses utama, sedangkan perulangan dalam digunakan untuk menjalankan proses yang lebih spesifik pada setiap iterasi.

Konsep ini sangat sesuai digunakan untuk membuat pola karena setiap baris dapat memiliki susunan karakter, spasi, dan angka yang berbeda.

### 5. Input dan Konversi Data

Fungsi input() digunakan untuk menerima data yang dimasukkan oleh pengguna melalui keyboard. Secara default, data yang diterima oleh input() memiliki tipe data string. Oleh karena itu, apabila nilai yang dimasukkan akan digunakan dalam operasi matematika atau perulangan, data tersebut perlu dikonversi menjadi integer menggunakan int().

### 6. Fungsi print() dan Parameter end

print() digunakan untuk menampilkan data ke layar. Secara default, print() akan berpindah ke baris berikutnya setelah mencetak suatu nilai. Parameter end dapat digunakan untuk mengubah perilaku tersebut.

Contohnya:

print("=", end="")

Perintah tersebut mencetak karakter = tanpa berpindah ke baris berikutnya. Penggunaan end="" sangat penting dalam pembuatan pola karena karakter harus dicetak secara berurutan pada baris yang sama.

Setelah seluruh bagian dalam satu baris selesai dicetak, perintah:

print()

digunakan untuk berpindah ke baris berikutnya.

### 7. Operator Perbandingan

Operator perbandingan digunakan untuk membandingkan dua nilai dan menghasilkan kondisi benar atau salah. Operator yang digunakan dalam program antara lain:

== untuk memeriksa kesamaan nilai. != untuk memeriksa ketidaksamaan nilai. > untuk memeriksa nilai lebih besar. < untuk memeriksa nilai lebih kecil. <= untuk memeriksa nilai lebih kecil atau sama dengan.

Operator perbandingan digunakan dalam percabangan untuk menentukan karakter atau proses yang harus dijalankan.

### 8. Operator Logika or dan and

Operator or digunakan ketika salah satu dari beberapa kondisi sudah cukup untuk memenuhi suatu kondisi. Operator and digunakan ketika beberapa kondisi harus terpenuhi secara bersamaan.

Dalam pola jam pasir, operator or digunakan untuk memeriksa posisi tertentu seperti baris pertama atau baris terakhir, sedangkan and digunakan untuk menentukan posisi karakter di bagian dalam pola.

### 9. Percabangan if, elif, dan else

Percabangan digunakan untuk menjalankan perintah berdasarkan kondisi tertentu. if digunakan untuk memeriksa kondisi pertama, elif digunakan untuk memeriksa kondisi tambahan, sedangkan else digunakan ketika kondisi sebelumnya tidak terpenuhi.

Pada program pola jam pasir, percabangan digunakan untuk menentukan karakter yang dicetak berdasarkan posisi baris dan kolom.

### 10. Pencetakan Pola

Pencetakan pola merupakan penerapan nested for untuk menghasilkan susunan karakter atau angka dalam bentuk tertentu. Pola dapat dibentuk dengan mengatur jumlah karakter, jumlah spasi, posisi angka, serta perpindahan baris.

Pada praktikum ini, pola yang dibuat meliputi pola titik dan angka, pola persegi menggunakan karakter *, pola hasil penjumlahan i + j, pola jam pasir, serta pola segitiga.

### 11. Pola Jam Pasir Kristal Palindromik

Pola jam pasir kristal palindromik menggunakan kombinasi perulangan, percabangan, perhitungan, dan pencetakan karakter. Program membentuk bagian atas, poros tengah, dan bagian bawah secara teratur.

Deret angka dibuat secara menurun dan kemudian menaik pada setiap baris. Simbol <*> digunakan sebagai poros kristal, sedangkan karakter #, =, dan | digunakan untuk membentuk bingkai pola.

Pengaturan jumlah spasi pada sisi kiri dan kanan digunakan untuk menjaga keseimbangan serta bentuk pola agar tetap simetris.

### 2.2.1 Program "perulangan_for1_2611532036"

```python
ulang_2036 = int(input("Masukkan jumlah perulangan: "))

for i in range(ulang_2036):
    print(f"Perulangan ke-{i}")
```

2. for i in range(ulang_2036):

- Digunakan untuk membuat proses perulangan menggunakan struktur for.

### 2.2.3 Program "perulangan_for3_2611532036"

```python
ulang_2036 = int(input("Masukkan jumlah perulangan: "))

jumlah_2036 = 0
for i in range(1, ulang_2036 + 1):
    print(i, end=" ")
    jumlah_2036 = jumlah_2036 + i

    if i < ulang_2036:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_2036, end=" ")
print()
print("jumlah = ", jumlah_2036)
```

5. Baris 6 berfungsi untuk menjumlahkan nilai i ke dalam variabel jumlah_2036. Hasil penjumlahan sebelumnya akan ditambahkan dengan nilai i pada setiap proses perulangan.
6. Baris 8 berfungsi untuk memeriksa apakah nilai i masih lebih kecil dari ulang_2036. Jika kondisi bernilai benar, program akan mencetak tanda + sebagai pemisah antarangka.

### 2.2.4 Program "nested_for1_2611532036"

```python
batas_2036 = int(input("Masukkan nilai batas: "))
for line_2036 in range(1, batas_2036 + 1):
    for j in range(1, (-1 * line_2036 + batas_2036) + 1):
        print(".", end=" ")
    print(line_2036)
```

3. Baris 3 berfungsi sebagai perulangan dalam (inner loop) yang digunakan untuk menentukan jumlah titik yang akan dicetak pada setiap baris. Rumus (-1 * line_2036 + batas_2036) + 1 menentukan jumlah pengulangan berdasarkan posisi baris.
4. Baris 4 berfungsi untuk mencetak karakter titik (.) pada setiap proses perulangan. Parameter end=" " digunakan agar setiap titik dicetak pada baris yang sama dan dipisahkan oleh spasi.
5. Baris 5 berfungsi untuk mencetak nilai line_2036 setelah seluruh titik pada baris tersebut selesai dicetak. Setelah itu, print() secara otomatis membuat baris baru.

### 2.2.6 Program "nested_for3_2611532036"

```python
batas_2036 = int(input("Masukkan nilai batas: "))
for i in range(batas_2036+1):
    for j in range(batas_2036+1):
        print(i+j, end=" ")
    print() # pindah ke baris berikutnya
```

3. Baris 3 berfungsi sebagai perulangan dalam (*inner loop*) menggunakan `range(batas_2036 + 1)`. Perulangan ini menentukan nilai `j` mulai dari `0` sampai `batas_2036` pada setiap baris.
4. Baris 4 berfungsi untuk mencetak hasil penjumlahan antara nilai `i` dan `j`. Parameter `end=" "` digunakan agar hasil penjumlahan dicetak pada baris yang sama dan dipisahkan oleh spasi.
5. Baris 5 berfungsi untuk berpindah ke baris berikutnya setelah seluruh proses perulangan dalam selesai. Fungsi `print()` tanpa argumen digunakan untuk membuat baris baru.

### 2.2.7 Program "nested_for4_2611532036"

```python
tinggi_2036 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2036 % 2 != 0:
    print("Tinggi pola harus bilangan genap!")
else:
    a_2036 = tinggi_2036
    c_2036 = a_2036
    lebar_2036 = (2 * tinggi_2036) - 2

    for i_2036 in range(1, tinggi_2036 + 1):
        b_2036 = c_2036 + 1

        for j_2036 in range(1, lebar_2036 + 1):

            # Baris atas dan bawah
            if i_2036 == 1 or i_2036 == tinggi_2036:
                if j_2036 == 1 or j_2036 == lebar_2036:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j_2036 == 1 or j_2036 == lebar_2036:
                    print("|", end="")
                else:
                    if j_2036 == c_2036:
                        print("<", end="")
                    elif j_2036 == b_2036:
                        print(">", end="")
                    elif j_2036 == (lebar_2036 - c_2036):
                        print("<", end="")
                    elif j_2036 == (lebar_2036 - c_2036 + 1):
                        print(">", end="")
                    elif j_2036 > b_2036 and j_2036 < (lebar_2036 - c_2036):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli Java
        a_2036 -= 2

        if a_2036 <= 0:
            c_2036 = (-a_2036) + 2
        else:
            c_2036 = a_2036
```

**Penjelasan:**
1. Baris 1 berfungsi untuk meminta pengguna memasukkan tinggi pola. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `tinggi_2036`.
2. Baris 3 digunakan untuk memeriksa apakah nilai `tinggi_2036` merupakan bilangan ganjil. Operator `%` digunakan untuk mendapatkan sisa hasil bagi. Jika hasilnya tidak sama dengan `0`, berarti nilai yang dimasukkan adalah bilangan ganjil.
3. Baris 4 berfungsi untuk menampilkan pesan bahwa tinggi pola harus berupa bilangan genap apabila input yang diberikan merupakan bilangan ganjil.
4. Baris 5 merupakan kondisi `else` yang dijalankan apabila nilai `tinggi_2036` merupakan bilangan genap.
5. Baris 6 digunakan untuk memberikan nilai `tinggi_2036` kepada variabel `a_2036`. Variabel ini digunakan sebagai nilai awal dalam menentukan posisi pola.
6. Baris 7 digunakan untuk memberikan nilai `a_2036` kepada variabel `c_2036`. Variabel `c_2036` digunakan untuk menentukan posisi karakter `<` dan `>` pada pola.
7. Baris 8 digunakan untuk menghitung lebar pola dengan rumus `(2 * tinggi_2036) - 2`. Hasil perhitungan disimpan dalam variabel `lebar_2036`.
8. Baris 10 merupakan perulangan luar (*outer loop*) yang digunakan untuk membentuk setiap baris pola. Fungsi `range(1, tinggi_2036 + 1)` membuat perulangan berjalan dari `1` sampai `tinggi_2036`.
9. Baris 11 digunakan untuk menentukan nilai `b_2036` dengan menambahkan `1` pada nilai `c_2036`. Nilai tersebut digunakan untuk menentukan salah satu posisi karakter `>`.
10. Baris 13 merupakan perulangan dalam (*inner loop*) yang digunakan untuk membentuk setiap karakter pada satu baris pola. Perulangan berjalan dari `1` sampai `lebar_2036`.
11. Baris 16 digunakan untuk memeriksa apakah baris yang sedang diproses merupakan baris pertama atau baris terakhir. Kondisi menggunakan operator `or` agar salah satu dari kedua kondisi tersebut sudah cukup untuk terpenuhi.
12. Baris 17 digunakan untuk memeriksa apakah posisi `j_2036` berada pada ujung kiri atau ujung kanan pola.
13. Baris 18 berfungsi untuk mencetak karakter `#` pada bagian sudut kiri dan kanan pola. `end=""` digunakan agar karakter berikutnya tetap dicetak pada baris yang sama.
14. Baris 20 berfungsi untuk mencetak karakter `=` pada bagian tengah bingkai atas dan bawah pola.

### 2.2.8 Program "jumlah_genap_2611532036"

```python
ulang_2036 = int(input("Masukkan nilai batas: "))

jumlah_2036 = 0
for i in range(1, ulang_2036 + 1):
    if i % 2 == 0:
        print(i, end=" ")
        jumlah_2036 = jumlah_2036 + i

        if i < ulang_2036:
            print(" + ", end=" ")
        else:
            print(" = ", jumlah_2036, end=" ")
print()
print("jumlah = ", jumlah_2036)
```

**Penjelasan:**
1. Baris 1 berfungsi untuk meminta pengguna memasukkan nilai batas. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `ulang_2036`.
2. Baris 3 digunakan untuk menginisialisasi variabel `jumlah_2036` dengan nilai `0`. Variabel ini digunakan untuk menyimpan hasil penjumlahan bilangan yang memenuhi kondisi.
3. Baris 4 merupakan perulangan `for` yang digunakan untuk memeriksa bilangan dari `1` sampai `ulang_2036`.
4. Baris 5 digunakan untuk memeriksa apakah nilai `i` merupakan bilangan genap. Operator `%` digunakan untuk mendapatkan sisa hasil bagi. Jika `i % 2 == 0`, berarti nilai `i` merupakan bilangan genap.
5. Baris 6 berfungsi untuk mencetak nilai `i` yang merupakan bilangan genap. `end=" "` digunakan agar nilai berikutnya tetap dicetak pada baris yang sama.
6. Baris 7 digunakan untuk menjumlahkan nilai `i` ke dalam variabel `jumlah_2036`. Hasil penjumlahan disimpan kembali ke dalam `jumlah_2036`.
7. Baris 9 digunakan untuk memeriksa apakah nilai `i` masih lebih kecil dari `ulang_2036`. Kondisi ini digunakan untuk menentukan apakah tanda `+` atau `=` yang akan ditampilkan.
8. Baris 10 berfungsi untuk mencetak tanda `+` apabila masih terdapat bilangan genap berikutnya. `end=" "` digunakan agar output tetap berada pada baris yang sama.
9. Baris 12 merupakan kondisi `else` yang dijalankan apabila nilai `i` sudah mencapai batas `ulang_2036`.
10. Baris 13 berfungsi untuk mencetak tanda `=` beserta hasil penjumlahan yang tersimpan dalam `jumlah_2036`.
11. Baris 14 digunakan untuk berpindah ke baris berikutnya setelah proses perulangan selesai.
12. Baris 15 berfungsi untuk menampilkan hasil akhir penjumlahan bilangan genap yang tersimpan dalam variabel `jumlah_2036`.

Berbeda dengan pola dasar, program ini menuntut pembentukan deret angka yang bergerak dinamis menurun dan menaik secara palindromik pada setiap baris serta diselaraskan dengan poros kristal di pusat bangun.

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

n_2036 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Batas Atas

print("#", end="")

for garis_2036 in range(4 * n_2036 + 5):
    print("=", end="")

print("#")

# Fase 1: Jam Pasir Atas

for baris_2036 in range(n_2036, 0, -1):

    print("| ", end="")

    # Spasi penyeimbang kiri
    for spasi_2036 in range(2 * (n_2036 - baris_2036)):
        print(" ", end="")

    # Deret angka menurun
    for angka_2036 in range(baris_2036, 0, -1):
        print(angka_2036, end=" ")

    # Poros kristal
    print("<*>", end="")

    # Deret angka menaik
    print(" ", end="")
    for angka_2036 in range(1, baris_2036 + 1):
        print(angka_2036, end=" ")

    # Spasi penyeimbang kanan
    for spasi_kanan_2036 in range(2 * (n_2036 - baris_2036)):
        print(" ", end="")

    print("|")

# Fase 2: Poros Titik Pusat Jam Pasir

print("|", end="")

for spasi_tengah_2036 in range(2 * n_2036 + 1):
    print(" ", end="")

print("<*>", end="")

for spasi_tengah_kanan_2036 in range(2 * n_2036 + 1):
    print(" ", end="")

print("|")

# Fase 3: Jam Pasir Bawah

for baris_bawah_2036 in range(1, n_2036 + 1):

    print("| ", end="")

    # Spasi penyeimbang kiri
    for spasi_bawah_2036 in range(2 * (n_2036 - baris_bawah_2036)):
        print(" ", end="")

    # Deret angka menurun
    for angka_bawah_2036 in range(baris_bawah_2036, 0, -1):
        print(angka_bawah_2036, end=" ")

    # Poros kristal
    print("<*>", end="")

    # Deret angka menaik
    print(" ", end="")
    for angka_naik_2036 in range(1, baris_bawah_2036 + 1):
        print(angka_naik_2036, end=" ")

    # Spasi penyeimbang kanan
    for spasi_kanan_bawah_2036 in range(2 * (n_2036 - baris_bawah_2036)):
        print(" ", end="")

    print("|")

# Batas Bawah

print("#", end="")

for garis_bawah_2036 in range(4 * n_2036 + 5):
    print("=", end="")

print("#")
```
